```python
import pandas as pd
```

```python
data_import = pd.read_csv('/content/drive/MyDrive/Exercises/data_cleaning_challenge.csv')
data_import
```

```python
drop_extra = data_import.drop(columns=['Unnamed: 9','Unnamed: 10'])
drop_extra
```

```python
drop_nans = drop_extra[drop_extra["Row Type"].notna()]
drop_nans
```

```python
column_values = []
counter = 0

for i in drop_nans['Row Type']:
  if "first name" in i:
    counter += 1
    column_values.append(counter)
```

```python
iter_cols = drop_nans.copy()

# Create a boolean series indicating where 'first name' is found
is_first_name = iter_cols['Row Type'].str.contains("first name", na=False)

# Create a series where a number is assigned only when 'first name' is found, otherwise NaN.
# The cumulative sum of the boolean series gives the iteration number for each 'first name' row.
iteration_series = is_first_name.cumsum().where(is_first_name)

# Forward-fill the NaN values with the last valid iteration number.
# This propagates the iteration number to all rows within a block.
iter_cols['Iteration'] = iteration_series.ffill()

iter_cols
```

```python
drop_extra_column_name_rows = iter_cols[iter_cols["Row Type"] != "Row Type"]
drop_extra_column_name_rows
```

```python
name_dataframe = drop_extra_column_name_rows[drop_extra_column_name_rows["Row Type"].str.contains('first name')]
name_dataframe.drop(columns=['Speed1','Speed2','Electricity','Effort','Weight','Torque'], inplace=True)
name_dataframe.rename(columns={'Row Type':'First Name', 'Iter Number':'Last Name','Power1':'Date'}, inplace=True)
name_dataframe['First Name'] = name_dataframe['First Name'].str[12:]
name_dataframe['Last Name'] = name_dataframe['Last Name'].str[11:]
name_dataframe['Date'] = name_dataframe['Date'].str[6:]
name_dataframe
```

```python
no_name_dataframe = drop_extra_column_name_rows[~drop_extra_column_name_rows["Row Type"].str.contains('first name')]
no_name_dataframe
```

```python
final_dataframe = pd.merge(left=name_dataframe, right=no_name_dataframe, how='inner', on='Iteration')
final_dataframe = final_dataframe.drop(columns=['Iteration'])
final_dataframe
```
